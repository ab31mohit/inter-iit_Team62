```python
import pandas as pd

def interactive_plotter_csv(csv_file):
    # Read the CSV file
    data = pd.read_csv(csv_file, header=None)
    data.columns = ['timestamp', 'u1', 'u2', 'f1', 'f2', 'f3', 'p_s', 'q_s', 'r_s', 'nx_des_frd_gnd',
                    'ny_des_frd_gnd', 'nz_des_frd_gnd', 'nx_des_frd_body', 'ny_des_frd_body', 'nz_des_frd_body', 
                    'n_x', 'n_y', 'n_z', 'roll_rad', 'pitch_rad', 'yaw_rad', 'x', 'y', 'z']
    data['timestamp'] = (data['timestamp'] - data['timestamp'].iloc[0]) / 1e6  # Convert to seconds
    # Plot each column against the timestamp
    for column in data.columns[1:]:  # Skip the first column (timestamp)
        plt.figure(figsize=(10, 5))
        plt.plot(data['timestamp'], data[column], label=column)
        plt.xlabel('Timestamp (s)')
        plt.ylabel('Values')
        plt.title(f'Odometry Data for {column} from file: {csv_file}')
        plt.legend()
        plt.grid()
        plt.show()

    # Compute and plot the sum of f1, f2, and f3
    data['f_sum'] = data['f1'] + data['f2'] + data['f3']
    plt.figure(figsize=(10, 5))
    plt.plot(data['timestamp'], data['f_sum'], label='f_sum')
    plt.xlabel('Timestamp (s)')
    plt.ylabel('Sum of f1, f2, f3')
    plt.title(f'Sum of f1, f2, and f3 from file number: {csv_file}')
    plt.legend()
    plt.grid()
    plt.show()
```

```python
import os
file_list = []
directory = '.'
for filename in os.listdir(directory):
    if filename.endswith('.csv') and filename[-5].isdigit():
        file_list.append(filename)
file_list.sort(key=lambda filename: int(filename[14:-4]))
```

```python
interactive_plotter_csv(file_list[-1])
```
